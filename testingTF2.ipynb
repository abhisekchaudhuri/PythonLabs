```python
import numpy as np
import tensorflow as tf
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense, InputLayer

#x = np.array([[200, 10], [150, 20], [100, 45]], dtype=np.float32)
#y = np.array([[1],
 #             [0],
 #             [1]])

x = np.array([

    [1.0, 0.05],

    [0.9, 0.06],

    [0.8, 0.08],

    [0.75,0.10],

    [0.6, 0.18],

    [0.5, 0.225]

])

y = np.array([

    [1],

    [1],

    [1],

    [0],

    [1],

    [1]

])

#scaling training data
#x=x/np.max(x)

#Define model
model=Sequential ([
    InputLayer(input_shape=(2,)),

    Dense(4, activation='relu'),       #hidden layer
    Dense(1, activation='sigmoid'),



])

print("Model is :", model.summary())

y_hat=model.predict(x)
print("y_hat is :", y_hat)
pred = model.predict(x)

for i in range(len(x)):

    print(

        f"Input={x[i]}, "

        f"Pred={pred[i][0]:.4f}, "

        f"Label={y[i][0]}"

    )

print("training model")


#Gradient descenr
model.compile(
    optimizer='adam',
    loss='binary_crossentropy'
)
#Training model via backward propagation
model.fit(x, y, epochs=100)
print("trained model", model.summary())
y_hat=model.predict(x)
print("y_hat is :", y_hat)
pred = model.predict(x)

for i in range(len(x)):

    print(

        f"Input={x[i]}, "

        f"Pred={pred[i][0]:.4f}, "

        f"Label={y[i][0]}"

    )
```
